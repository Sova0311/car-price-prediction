# EDA – Predikcija cene polovnih automobila

Analiza skupa `cars.csv` pre čišćenja i treniranja modela.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/cars.csv')
```

## Osnovne informacije

```python
print('Dimenzije:', df.shape)
display(df.head())
df.info()
```

## Nedostajuće vrednosti i statistika

```python
display(df.isna().sum().sort_values(ascending=False).to_frame('missing'))
display(df.describe(include='all').T)
```

## Distribucija ciljne promenljive

```python
fig, ax = plt.subplots(figsize=(9, 4))
df['priceUSD'].hist(bins=60, ax=ax)
ax.set_title('Distribucija cena automobila')
ax.set_xlabel('Cena (USD)')
ax.set_ylabel('Broj oglasa')
plt.show()
```

## Cena u odnosu na godište i kilometražu

```python
fig, ax = plt.subplots(figsize=(8, 4))
ax.scatter(df['year'], df['priceUSD'], alpha=.15)
ax.set_xlabel('Godina proizvodnje')
ax.set_ylabel('Cena (USD)')
ax.set_title('Cena prema godištu')
plt.show()

fig, ax = plt.subplots(figsize=(8, 4))
ax.scatter(df['mileage(kilometers)'], df['priceUSD'], alpha=.15)
ax.set_xlabel('Kilometraža (km)')
ax.set_ylabel('Cena (USD)')
ax.set_title('Cena prema kilometraži')
plt.show()
```
